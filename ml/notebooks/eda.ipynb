# Multimodal Fraud Detection: Exploratory Data Analysis & Benchmark Report

This notebook documents exploratory data analysis on the two core datasets:
1. **SMS Dataset** (`Dataset_5971.csv` / `sms_mendeley.csv`): 5,971 raw SMS records with class breakdown: Ham, Spam, Smishing.
2. **PhiUSIIL Phishing URL Dataset** (`Dataset_5971.csv` / `PhiUSIIL_Phishing_URL_Dataset.csv`): 235,795 URLs with `label` (1 = Legitimate: 134,850, 0 = Phishing: 100,945).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
from pathlib import Path

sns.set_theme(style="darkgrid")
print("Libraries imported successfully.")
```

## 1. SMS Phishing Corpus Inspection

```python
sms_path = Path("../datasets/processed/cleaned_sms.csv")
if sms_path.exists():
    df_sms = pd.read_csv(sms_path)
    print("Cleaned SMS Dataset Shape:", df_sms.shape)
    print("\nClass Distribution:")
    print(df_sms['label'].value_counts())
    print("\nSample Messages:")
    print(df_sms[['message_text', 'label', 'extracted_url']].head(5))
```

## 2. PhiUSIIL Phishing URL Dataset Inspection (Lexical Feature Subset)

```python
url_path = Path("../datasets/processed/cleaned_url.csv")
if url_path.exists():
    df_url = pd.read_csv(url_path, nrows=10000)
    print("URL Sample Shape:", df_url.shape)
    print("\nURL Label Distribution (1=Fraud, 0=Legitimate):")
    print(df_url['is_fraud'].value_counts())
    print("\nMean Lexical Features by Class:")
    print(df_url.groupby('label')[['URLLength', 'NoOfSubDomain', 'IsHTTPS', 'Entropy', 'SpacialCharRatioInURL']].mean())
```

## 3. Multimodal Pairing & Brand Consistency Augmentation

```python
pairs_path = Path("../datasets/processed/merged_pairs.csv")
if pairs_path.exists():
    df_pairs = pd.read_csv(pairs_path)
    print("Paired Dataset Shape:", df_pairs.shape)
    print("\nPair Type Distribution:")
    print(df_pairs['pair_type'].value_counts())
    print("\nLabels in Organic Pairs:")
    print(df_pairs[df_pairs['pair_type'] == 'organic']['label'].value_counts())
```

## 4. Benchmark & Ablation Study Results

```python
results_path = Path("../experiments/results/evaluation_summary.json")
if results_path.exists():
    with open(results_path, "r", encoding="utf-8") as f:
        eval_summary = json.load(f)
    print("\n=== Model Evaluation Summary ===")
    for m_name, m_stats in eval_summary.get('models', {}).items():
        print(f"{m_name:15s} -> Accuracy: {m_stats.get('accuracy')}, ROC-AUC: {m_stats.get('roc_auc')}, F1: {m_stats.get('f1_score')}")
```
